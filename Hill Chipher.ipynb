```python
import numpy as np
# 1. Korespondenkan abjad: A=1 ... Z=26
def text_to_numbers(text):
    return [ord(c.upper()) - ord('A') + 1 for c in text if c.isalpha()]
def numbers_to_text(numbers):
    return ''.join([chr((n - 1) % 26 + ord('A')) for n in numbers])
# Invers modulo biasa
def modinv(a, m):
    a = a % m
    for x in range(1, m):
        if (a * x) % m == 1:
            return x
    raise Exception("Tidak ada invers modul")
# Invers matriks modulo 26
def matrix_mod_inverse(matrix, modulus):
    det = int(round(np.linalg.det(matrix)))
    det = det % modulus
    det_inv = modinv(det, modulus)
    # Adjugate matrix
    adj = np.round(det * np.linalg.inv(matrix)).astype(int) % modulus
    inv_matrix = (det_inv * adj) % modulus
    return inv_matrix
# Algoritma Enkripsi Hill Cipher 
def hill_encrypt(plaintext, key_matrix):
    n = key_matrix.shape[0]
    numbers = text_to_numbers(plaintext)    
    while len(numbers) % n != 0:
        numbers.append(24)  # 'X' = 24 jika A=1
    blocks = np.array(numbers).reshape(-1, n).T  # transpose blok P
    encrypted_blocks = np.dot(key_matrix, blocks) % 26
    encrypted_blocks[encrypted_blocks == 0] = 26  # ubah 0 ke 26 (karena A=1, Z=26)
    encrypted_blocks = encrypted_blocks.T.flatten()
    return numbers_to_text(encrypted_blocks)
# Dekripsi sesuai asumsi
def hill_decrypt(ciphertext, key_matrix):
    n = key_matrix.shape[0]
    numbers = text_to_numbers(ciphertext)
    blocks = np.array(numbers).reshape(-1, n).T
    inv_key = matrix_mod_inverse(key_matrix, 26)
    decrypted_blocks = np.dot(inv_key, blocks) % 26
    decrypted_blocks[decrypted_blocks == 0] = 26
    decrypted_blocks = decrypted_blocks.T.flatten()
    return numbers_to_text(decrypted_blocks)
# Contoh penggunaan
key = np.array([[3, 3], [2, 5]]) 
plaintext = "TOLONG"
cipher = hill_encrypt(plaintext, key)
print("Ciphertext :", cipher)
decrypted = hill_decrypt(cipher, key)
print("Decrypted  :", decrypted)
```
